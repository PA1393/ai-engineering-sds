### Import the libraries and load the environment variables

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()
OPENAPI_API_KEY = os.getenv("OPENAPI_API_KEY")

client = OpenAI()
```

### Step 2: Call the API

```python
response = client.chat.completions.create(
    model="gpt-4o",
    messages=[
        {"role": "user", 
        "content": "Help me?"},
    ],
)

print(response)
```

```python
answer = response.choices[0].message.content
print(answer)
```

### Step 3a: Calling API with System Prompt

```python
response = client.chat.completions.create(
    model="gpt-4o",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."}, #System prompt
        {"role": "user", "content": "What projects could i build to break into NeuroTech as someone who has a background in software engineering but no experience in neuroscience? 6 months of software exp"}, #User prompt
    ],
)

answer = response.choices[0].message.content
display(Markdown(answer))
```

### IKEA Customer Service Agent

```python
question = "I bought a chair from your store last week and it's already broken. Can I return itf"
```

```python
response = client.chat.completions.create(
    model="gpt-4o",
    messages=[
        {"role": "system", "content": "You are an IKEA customer service representative assistant. But you strictly follow the return policy of IKEA allow a 365 day return policy."}, #System prompt
        {"role": "user", "content": question},]
)
answer = response.choices[0].message.content
print(answer)
```
