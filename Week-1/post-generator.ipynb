### Import libraries and env vars

```python
from openai import OpenAI
from dotenv import load_dotenv
import os
from IPython.display import display, Markdown
from pprint import pprint

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
client = OpenAI()
```

### Web Searching Model

```python
response = client.chat.completions.create(
    model="gpt-4o-mini-search-preview",
    messages=[
        {"role":"user", "content": "Write me a article including project details about AI detection tools for Parkinson disease using EEG sleep waveform analysis. Use reliable sources from Neuroscience credible journals (and .gov website data) to find real data for this post."}],
    max_tokens=500,
)

article_content = response.choices[0].message.content

display(Markdown(article_content))
```

### Summarizer + LinkedIn Creator Agent

```python
system_message = """ 
You are a helpful assistant who creates amazing linkedIn posts about AI in healthcare.
The post should be engaging and informative, and should include a call to action for readers to learn more about the topic.
Also make conscise and add a thought provoking question at the end of the post to encourage engagement.
"""

prompt = f"""
The following contains content of an article about AI detection tools for Parkinson disease using EEG sleep waveform analysis. Now please write
a linkedIn post about this article. 

{article_content}
"""

completion = client.chat.completions.create(
    model = "gpt-4o-mini", #summarizer model
    messages = [
        {"role": "system", "content": system_message},
        {"role": "user", "content": prompt},
    ],
    max_tokens=300,
    temperature=1.5,

)

post_content = completion.choices[0].message.content
display(Markdown(post_content)) 
```
