### Import libraries and load env variables

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()

OPENAPI_API_KEY = os.getenv("OPENAPI_API_KEY")

client = OpenAI()
```

### Step 2: Call API

```python

response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[
        {"role": "system", "content": "You are a helpful assistant "},
        {"role": "user", "content": "Are james coudis action halal?"},
    ],
)   

answer = response.choices[0].message.content
display(Markdown(answer))
```

### Context Window (how it works in theory)

```python
conv_history = [
    {"role": "system", "content": "You are a helpful assistant "},
    {"role": "user", "content": "I am travelling to Hawaii, what are some good places to visit?"},
    {"role": "assistant", "content": "Some good places to visit in Hawaii include Waikiki Beach, Pearl Harbor, and the Hawaii Volcanoes National Park."},
]

response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=conv_history +
    [
        {"role": "user", "content": "Where am I travelling to?"},
    ]
)

answer = response.choices[0].message.content
print(answer)
```

### Appendling LLM response to list

```python
from pprint import pprint   

conv_history = [
    {"role": "system", "content": "You are a helpful assistant "},
    {"role": "user", "content": "I am travelling to Hawaii, what are some good places to visit?"},
    {"role": "assistant", "content": "Some good places to visit in Hawaii include Waikiki Beach, Pearl Harbor, and the Hawaii Volcanoes National Park."},
]
prompt = {"role": "user", "content": "Where am I travelling to? and why"}
conv_history.append(prompt)

response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=conv_history 
)

answer = response.choices[0].message.content

replyToAppend = {"role": "assistant", "content": answer}
conv_history.append(replyToAppend)

pprint(conv_history)
```
