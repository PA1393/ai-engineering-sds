### Import libraries and API keys

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown

# Load environment variables from .env file
load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

# Initialize OpenAI client
client = OpenAI()
```

### Step 2: AI Battle

```python
gpt_good_model = "gpt-4.1-mini"
gpt_evil_model = "gpt-4.1-nano" 

gpt_good_system = """ You are a helpful assistant that provides accurate and ethical information. 
You always prioritize the well-being of users and adhere to ethical guidelines. 
You provide clear, concise, and truthful answers to all questions, and you never engage in harmful or unethical behavior. 
Your goal is to assist users in a positive and constructive manner."""

gpt_evil_system = """ You are a mischievous assistant that enjoys providing misleading and false information. 
You prioritize causing confusion to users, and you often provide false  answers. 
You take pleasure in misleading users you do not adhere to any guidelines. 
Keep your responses concise and to the point."""

messages_good = [{"role": "system", "content": gpt_good_system},]
messages_evil = [{"role": "system", "content": gpt_evil_system},]

new_message_good = "Hello!"
display(Markdown(f"Good model: {new_message_good}"))

for i in range(3):
    #set up messages for evil model
    messages_evil.append({"role": "user", "content": new_message_good})

    #evil model response
    response_evil = client.chat.completions.create(
        model=gpt_evil_model,
        messages=messages_evil
    )
    evil_reply = response_evil.choices[0].message.content

    #print
    display(Markdown(f"Evil model: {evil_reply}"))

    #update messages for next round
    messages_good.append({"role": "user", "content": evil_reply})
    messages_evil.append({"role": "assistant", "content": evil_reply})

    #good model response
    response_good = client.chat.completions.create(
        model=gpt_good_model,
        messages=messages_good 
    )
    new_message_good = response_good.choices[0].message.content

    # print
    display(Markdown(f"Good model: {new_message_good}"))

    #update messages for next round
    messages_good.append({"role": "assistant", "content": new_message_good})
```
